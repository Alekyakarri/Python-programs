```python
# program that finds Maximum among three numbers
a=int(input("Enter a number"))
b=int(input("Enter number b"))
c=int(input("Enter number c"))
if(a>b and a>c):
  print(f"a is greater nunumber")
elif(b>a and b>c):
  print(f"b is greater number")
else:
  print(f"c is greater number")
```

```python
# program for given year is leap year or not
year=int(input("Enter year"))
if(year%4==0 and year%100!=0 or year%400==0):
  print(f"is leap year")
else:
  print(f"is not a leap year")
```

```python
#program using match case
day=int(input("Enter day"))
match day:
  case 1 :
    print("Sunday")
  case 2:
    print("Monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print("Thursday")
  case 6:
    print("Friday")
  case 7:
    print("Saturday")
  case_:
    print("Invalid")
```
