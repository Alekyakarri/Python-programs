```python
c = {}
print(type(c))
# syntax
d = {"s.no":210,"name":"Alekya","grade":9.9,"marks":99}
print(d)
# using dict keyword
d1= dict(((1,2),("s.no",10),(4,5)))
print(d1)
print(len(d))
print(max(d))
print(min(d))
print(d["name"]) # accessing element
d.pop("name") # removing
print(d)
print(sorted(d))
d.update({"age":21,"height":5.9})
print(d)
```

```python
# iteration
for i in d:    # keys
  print(i)
for i in d:    # values
  print(d[i])
for i in d :
  print(i,":",d[i])
```

```python
#separating keys
print(d.keys())
a = list(d.keys())
print(a)
print(type(a))
b = list(d.values())
print(b)
print(type(b))
c = list(d.items())
print(b)
for k,v in c:
  print(k,v)
```

```python
# Dictationary comprehension
d = {i:i*i for i in range(1,6) }
print(d)
# User defined dictionary
k = [i for i in input().split()]
v = [int(i) for i in input().split()]
d = {i:j for i,j in zip(k,v)}
print(d)
```
