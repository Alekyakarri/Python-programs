```python
# performing operations on tuple
# creating an empty tuple
t=()
print(t)
print(type(t))
#operations on tuple
t1=(1,2,3,4,'ram','sita')
t2=(5,'laxmi')
#accessing element or indexing
print(t1[0])
print(t1[-1])
#concatenation operator
print(t1+t2)
#repitition operator
print(t1*3)
#slicing operator
print(t1[0:3])
#membership operator
print(5 in t2)
print('ramya' in t1)




```

```python
#functions in tuple
#length in tuple
t3=(18,28,38,48,'Alekya',48,'Alekya')
print(len(t3))
print(t3.count(48))
print(t3.index('Alekya'))
print(t3.index(18))
```

```python
# Zip operation
print(tuple(zip(t2,t3)))
```
