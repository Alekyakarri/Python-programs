```python
# Sets in python
S=set()
print(type(S))
#set operations
s1={1,2,3,4}
s2={3,4,5,6}
print(s1|s2)
print(s1&s2)
print(s1-s2)
print(s1^s2)
```

```python
#Set methods
s3={2,4,6,8,12}
s3.add(10)
print(s3)
s3.update([3,4,5])
print(s3)
s3.discard(3)
print(s3)
s3.remove(5)
print(s3)
s4={66,77,88}
x=s4.pop()
print(x)
s5={55,44,33}
s5.clear()
print(s5)
print(s1.union(s2))
print(s3.intersection(s4))
print(s3.difference(s4))
print(s3.symmetric_difference(s4))
```
