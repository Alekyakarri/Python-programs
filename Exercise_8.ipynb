```python
# to find  no of duplicate from a list
l1 = [int(i) for i in input().split()]
S = set(l1)
print(S)
print(len(l1)-len(S))
```

```python
#  to find the count of no.of duplicates without loop
l1 = [int(i) for i in input().split()]
l2 = []
for i in l1:
  if l1.count(i) > 1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(s))

```

```python
# wrirte a python program print non repeted elements
l1 = [int(i) for i in input().split()]
l2 = []
for i in l1:
  if l1.count(i) == 1:
    l2.append(i)
print(l2)
```
