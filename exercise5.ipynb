```python
# program that reverse a string using function
def rev_str(S):
  string=S[::-1]
  return string
print(rev_str("abcd"))

```

```python
# using for loop user de
def rev_str(S):
  x=""
  for i in range(len(S)-1,-1,-1):
    x=x+S[i]
  return x
print(rev_str("hello"))
```

```python
# example code on default arguments
def print_msg(name,branch="AI&DS"):
  print("the student name is",name,"and the branch is",branch)
print_msg("ALEKYA")
```

```python
#example code on keyword arguments
def subtraction(a,b,c):
  return a-b-c
res=subtraction(a=1,b=2,c=3) #we can take variables irrespective of given order
print(res)
```

```python
# palindrome checking using function
def palindrome(S):
  S=S[::-1]
  if palindrome:
    print("palindrome")
  else:
    print("not a palindrome")
palindrome("madam")
```

```python
#function that returns multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_function(1,2,3,4,5,6,7,8)
print(f"the sum of numbers is {res_sum},the count is {res_count},the average is {res_avg}")
```
